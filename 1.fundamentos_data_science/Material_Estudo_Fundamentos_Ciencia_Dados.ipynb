## 1. O que é Data Analytics?
Data Analytics é o processo de analisar dados brutos para extrair informações significativas, identificar padrões e apoiar a tomada de decisões. 

Existem quatro tipos principais de análise de dados:
- **Análise Descritiva:** O que aconteceu? (Ex: Média de vendas, número de acessos).
- **Análise Diagnóstica:** Por que aconteceu? (Ex: Motivos para queda de vendas, causas da evasão escolar).
- **Análise Preditiva:** O que vai acontecer no futuro? (Ex: Prever demanda, prever *churn* de clientes). Utiliza machine learning e estatística.
- **Análise Prescritiva:** O que devemos fazer? (Ex: Otimizar estoque, recomendar rotas no Google Maps).

## 2. Python vs. C
Muitos estudantes de computação aprendem **C** nas disciplinas iniciais e depois migram para **Python** para Ciência de Dados. Aqui estão as principais diferenças:

| Característica | C | Python |
| :--- | :--- | :--- |
| **Nível** | Baixo/Médio (Próximo ao hardware) | Alto (Foco na legibilidade e usabilidade) |
| **Execução** | Compilada (gera executável) | Interpretada (executa linha a linha) |
| **Tipagem** | Estática (exige declaração de tipo) | Dinâmica (o tipo se adapta ao valor) |
| **Memória** | Gerenciamento manual (alocar/liberar) | Automático (*Garbage Collection*) |
| **Sintaxe** | Rígida (chaves, ponto e vírgula) | Simples (indentação obrigatória) |
| **Uso Principal**| Sistemas embarcados, drivers | Ciência de Dados, IA, Web, Automação |

```python
# Exemplo de Tipagem Dinâmica e Sintaxe Simples em Python

# No Python, não precisamos declarar o tipo da variável.
variavel = 10
print(f"Valor numérico: {variavel} | Tipo: {type(variavel)}")

# Podemos mudar o tipo livremente na mesma variável (Tipagem Dinâmica)
variavel = "Agora sou uma string!"
print(f"Valor texto: {variavel} | Tipo: {type(variavel)}")
```

### Exemplo Prático: Análise Descritiva com Pandas
Vamos simular um pequeno conjunto de dados de notas e analisar descritivamente:

```python
import pandas as pd

# Simulação de dados: Notas de alunos
dados = {
    'Aluno': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Nota': [8.5, 7.0, 9.2, 5.5, 8.0],
    'Status': ['Aprovado', 'Aprovado', 'Aprovado', 'Reprovado', 'Aprovado']
}

df = pd.DataFrame(dados)
display(df)

# Análise Descritiva simples
media = df['Nota'].mean()
mediana = df['Nota'].median()

print(f"\nMédia da turma: {media:.2f}")
print(f"Mediana da turma: {mediana:.2f}")
```

## 4. Classificação vs Regressão na Análise Preditiva
A professora mencionou uma distinção importante na análise preditiva (Machine Learning):
- **Regressão:** Prevê um valor contínuo numérico (Ex: Qual será o faturamento do mês que vem?).
- **Classificação:** Prevê uma categoria/rótulo (Ex: O cliente vai cancelar a assinatura - *Churn*? Sim ou Não? Que dígito é este no CEP?).
