## 1. Data Science Methodology Overview

### 1.1 O que é uma metodologia?
Uma **metodologia** é um sistema de métodos, princípios e regras que orienta a execução de um trabalho. Em Ciência de Dados ela garante que o trabalho seja:

- **Repetível** – outro profissional pode reproduzir os passos;
- **Rastreável** – cada decisão tem justificativa registrada;
- **Alinhado ao negócio** – o resultado responde a uma pergunta real;
- **Iterativo** – permite voltar a etapas anteriores quando surgem novas informações.

> "A metodologia é independente de tecnologia. Ferramentas mudam; a forma de pensar o problema permanece." — John Rollins (IBM)

### 1.2 As 10 etapas da metodologia (IBM Foundational Methodology)

A metodologia é organizada em **5 módulos** que agrupam **10 etapas**:

| Módulo | Etapa | Pergunta central |
|---|---|---|
| **Do problema à abordagem** | 1. Business Understanding | *Qual problema estamos tentando resolver?* |
| | 2. Analytic Approach | *Como podemos usar dados para responder à pergunta?* |
| **Dos requisitos à coleta** | 3. Data Requirements | *De quais dados precisamos?* |
| | 4. Data Collection | *De onde vêm os dados e como obtê-los?* |
| **Do entendimento à preparação** | 5. Data Understanding | *Os dados coletados representam o problema?* |
| | 6. Data Preparation | *Que trabalho adicional é preciso para manipular os dados?* |
| **Da modelagem à avaliação** | 7. Modeling | *De que forma os dados podem ser visualizados/modelados para obter a resposta?* |
| | 8. Evaluation | *O modelo realmente responde à pergunta?* |
| **Da implantação ao feedback** | 9. Deployment | *Como colocar a resposta em uso?* |
| | 10. Feedback | *O modelo continua respondendo bem com o passar do tempo?* |

### 1.3 Natureza iterativa
O fluxo **não é linear**. Alguns retornos comuns:

- *Data Understanding → Data Requirements*: descobrimos que faltam dados.
- *Evaluation → Modeling*: o modelo não atinge o desempenho exigido.
- *Feedback → Business Understanding*: o problema de negócio mudou.

```
Business Understanding ─► Analytic Approach ─► Data Requirements ─► Data Collection
        ▲                                                                 │
        │                                                                 ▼
    Feedback ◄─ Deployment ◄─ Evaluation ◄─ Modeling ◄─ Data Preparation ◄─ Data Understanding
```

### 1.4 Comparação com o CRISP-DM

| CRISP-DM (6 fases) | IBM Foundational Methodology (10 etapas) |
|---|---|
| Business Understanding | Business Understanding + Analytic Approach |
| Data Understanding | Data Requirements + Data Collection + Data Understanding |
| Data Preparation | Data Preparation |
| Modeling | Modeling |
| Evaluation | Evaluation |
| Deployment | Deployment + **Feedback** |

A principal contribuição da metodologia IBM é destacar explicitamente **Analytic Approach** (escolha da técnica antes de olhar os dados) e **Feedback** (melhoria contínua após a implantação).

## 2. Business Understanding

### 2.1 Definição
É a etapa que **define o problema** a ser resolvido e os **objetivos** do projeto do ponto de vista do negócio. Sem um entendimento claro, todas as etapas seguintes ficam comprometidas: *"garbage in, garbage out"* começa aqui.

### 2.2 Objetivos da etapa
1. Entender o **contexto** do negócio (setor, processos, restrições).
2. Identificar os **stakeholders** (patrocinador, usuários finais, área de TI, jurídico).
3. Definir **objetivos de negócio** (ex.: reduzir custos, aumentar retenção).
4. Traduzir objetivos em **critérios de sucesso mensuráveis**.
5. Levantar **restrições** (prazo, orçamento, regulamentações como LGPD).

### 2.3 Técnicas úteis
- **Entrevistas com stakeholders** e workshops de descoberta;
- **5 Porquês** para chegar à causa raiz;
- **Objetivos SMART** (Específico, Mensurável, Atingível, Relevante, Temporal);
- **Mapa de stakeholders** (poder × interesse);
- **Business Model Canvas / Problem Statement Canvas**.

### 2.4 Estudo de caso clássico (IBM): Reinternação hospitalar
Um hospital americano percebeu que pacientes com **insuficiência cardíaca congestiva (ICC)** retornavam ao hospital em menos de 30 dias após a alta. As reinternações geravam multas do sistema de saúde e altos custos.

- **Objetivo de negócio:** reduzir reinternações de pacientes com ICC.
- **Meta mensurável:** reduzir em X% a taxa de reinternação em 30 dias.
- **Stakeholders:** diretoria clínica, médicos, enfermagem, financeiro, TI.
- **Restrição:** decisões devem ser explicáveis aos médicos.

### 2.5 Entregáveis
- Documento de definição do problema (*Problem Statement*);
- Lista de perguntas de negócio priorizadas;
- Critérios de sucesso e KPIs;
- Glossário de termos do domínio.

## 3. Analytic Approach

### 3.1 Definição
Uma vez definido o problema, é preciso decidir **qual tipo de análise** responderá à pergunta. O *Analytic Approach* é a **ponte entre a pergunta de negócio e a técnica analítica**.

A escolha é feita **antes de olhar os dados em detalhe**, pois ela determina *quais* dados serão necessários (etapa 3).

### 3.2 Os 4 níveis de análise

| Nível | Pergunta que responde | Exemplo | Técnicas típicas |
|---|---|---|---|
| **Descritiva** | *O que aconteceu?* | Qual foi a taxa de reinternação em 2025? | Estatística descritiva, dashboards, agregações |
| **Diagnóstica** | *Por que aconteceu?* | Por que pacientes idosos reinternam mais? | Correlação, análise de causa raiz, testes de hipótese |
| **Preditiva** | *O que vai acontecer?* | Qual paciente tem alta probabilidade de reinternar? | Classificação, regressão, séries temporais |
| **Prescritiva** | *O que devemos fazer?* | Qual intervenção reduz a reinternação para cada perfil? | Otimização, simulação, sistemas de recomendação |

### 3.3 Famílias de técnicas
- **Estatística clássica** – quando se busca *inferência* (intervalos de confiança, testes).
- **Aprendizado supervisionado** – há uma variável alvo (classificação/regressão).
- **Aprendizado não supervisionado** – busca de estrutura (clustering, redução de dimensionalidade, regras de associação).
- **Séries temporais** – dados ordenados no tempo (ARIMA, Prophet, LSTM).
- **Otimização/Pesquisa Operacional** – decisão prescritiva.

### 3.4 Critérios de escolha
1. **Tipo de pergunta** (seção 4);
2. **Interpretabilidade exigida** – médicos preferem árvores de decisão a redes neurais;
3. **Volume e tipo de dados disponíveis**;
4. **Tempo de resposta** – batch ou tempo real;
5. **Maturidade da organização**.

## 4. Analytic Approach Based on the Question Type

A regra prática: **a forma da pergunta determina a forma da resposta**.

| Se a pergunta for… | Tipo de resposta | Abordagem analítica | Algoritmos exemplo |
|---|---|---|---|
| "Quantos / qual a média / qual a proporção?" | Número/resumo | **Estatística descritiva** | média, mediana, histograma |
| "Existe relação entre X e Y?" | Grau de associação | **Correlação / regressão / teste de hipótese** | Pearson, qui-quadrado, regressão linear |
| "Sim ou não? A ou B?" | Categoria binária | **Classificação binária** | Regressão logística, árvore de decisão, SVM |
| "Qual das N categorias?" | Categoria múltipla | **Classificação multiclasse** | Random Forest, XGBoost, redes neurais |
| "Quanto? Qual valor?" | Número contínuo | **Regressão** | Regressão linear, gradient boosting |
| "Quais grupos naturais existem?" | Segmentos | **Clustering** | K-Means, DBSCAN, hierárquico |
| "O que costuma ocorrer junto?" | Regras | **Regras de associação** | Apriori, FP-Growth |
| "Qual será o valor no futuro?" | Previsão temporal | **Séries temporais** | ARIMA, Prophet |
| "Isto é normal ou anômalo?" | Flag de anomalia | **Detecção de anomalias** | Isolation Forest, z-score |
| "Qual a melhor ação/alocação?" | Decisão | **Otimização / prescritiva** | Programação linear, simulação |
| "O que este texto/imagem diz?" | Estrutura extraída | **NLP / Visão computacional** | Transformers, CNNs |

### 4.1 Caso IBM: por que árvore de decisão?
No caso da ICC, a pergunta era: *"Este paciente tem alta probabilidade de ser reinternado em 30 dias? (sim/não)"* → **classificação binária**.

Entre os algoritmos possíveis, escolheu-se a **árvore de decisão** porque:
- produz **regras legíveis** ("se idade > 70 e nº de internações anteriores > 2 → alto risco");
- permite **ajustar o limiar de risco** conforme a capacidade da equipe de intervenção;
- lida bem com variáveis numéricas e categóricas sem muito pré-processamento.

````python
# Função auxiliar: sugerir abordagem a partir do tipo de pergunta
def sugerir_abordagem(pergunta: str) -> str:
    p = pergunta.lower()
    regras = [
        (["sim ou não", "vai ", "irá", "probabilidade de"], "Classificação (supervisionado)"),
        (["quanto", "qual valor", "quantos reais"],          "Regressão (supervisionado)"),
        (["grupos", "segmentos", "perfis"],                  "Clustering (não supervisionado)"),
        (["junto", "combinação", "cesta"],                   "Regras de associação"),
        (["próximo mês", "próximo ano", "tendência", "futuro"], "Séries temporais"),
        (["anômal", "fraude", "fora do padrão"],             "Detecção de anomalias"),
        (["relação", "influencia", "impacto", "por que"],    "Análise diagnóstica / correlação"),
        (["melhor", "otimizar", "alocar"],                   "Prescritiva / otimização"),
        (["quantos", "média", "qual foi", "proporção"],      "Estatística descritiva"),
    ]
    for chaves, abordagem in regras:
        if any(k in p for k in chaves):
            return abordagem
    return "Indefinido – refine a pergunta com o negócio"

perguntas = [
    "Este paciente vai reinternar em 30 dias?",
    "Quanto vamos faturar no próximo mês?",
    "Quais perfis de clientes existem na nossa base?",
    "Qual foi a taxa de churn em 2025?",
    "Que produtos são comprados juntos?",
    "Esta transação parece fraude?",
]
for q in perguntas:
    print(f"{q:<50} -> {sugerir_abordagem(q)}")
````

### 5.2 Estudo de caso completo: Reinternação por ICC

| Passo | Resultado no caso |
|---|---|
| **Objetivo de negócio** | Reduzir custos e multas com reinternações em 30 dias |
| **Pergunta de negócio** | Quais pacientes têm maior risco de reinternar? O que explica esse risco? |
| **Pergunta analítica** | Dado o perfil clínico e demográfico na alta, o paciente será reinternado em ≤ 30 dias? |
| **Tipo de resposta** | Sim/Não + probabilidade + fatores de risco |
| **Abordagem** | Classificação binária, modelo interpretável |
| **Técnica escolhida** | Árvore de decisão (posteriormente comparada com regressão logística) |
| **Critério de sucesso** | Sensibilidade (recall) alta para não deixar pacientes de risco sem acompanhamento |
| **Uso da resposta** | Enfermagem faz follow-up telefônico/visita nos pacientes de alto risco |

### 5.3 Armadilhas comuns
- **Começar pela técnica** ("quero usar deep learning") em vez da pergunta.
- **Pergunta vaga**: "melhorar a saúde dos pacientes" não é analisável.
- **Ignorar a ação**: se ninguém vai agir com a predição, o modelo não tem valor.
- **Métrica errada**: acurácia em dados desbalanceados esconde o problema (95% de acurácia prevendo "não reinterna" para todos).
- **Não envolver o usuário final** (médico, enfermeiro) na definição.

```
queremos responder [pergunta analítica]
usando [abordagem analítica]
para que [ação concreta] seja tomada,
medido por [KPI/critério de sucesso].
```

**Exemplo preenchido:**

## 6. Business Understanding: Identifying Relevant Questions

### 6.1 Da dor à pergunta
Stakeholders raramente chegam com perguntas prontas; chegam com **dores** ("estamos perdendo clientes"). O cientista de dados deve transformá-las em **perguntas relevantes e analisáveis**.

### 6.2 Técnica dos 5 Porquês (exemplo)
1. *Por que estamos perdendo receita?* → Clientes estão cancelando.
2. *Por que cancelam?* → Reclamam de atendimento lento.
3. *Por que o atendimento é lento?* → Fila concentrada em certos horários.
4. *Por que há concentração?* → Não prevemos a demanda por horário.
5. *Por que não prevemos?* → Não há modelo de previsão de demanda.

→ **Pergunta relevante:** "Qual será o volume de chamados por hora na próxima semana?" (séries temporais)

### 6.3 Critérios de uma boa pergunta (checklist)
- [ ] Está ligada a um **objetivo de negócio** claro?
- [ ] Alguém vai **agir** com a resposta?
- [ ] É **específica** (quem, o quê, quando, onde)?
- [ ] É **mensurável**?
- [ ] É **viável** com dados que existem ou podem ser coletados?
- [ ] Respeita **ética e regulamentação** (LGPD, viés)?

### 6.4 Priorização de perguntas (matriz valor × esforço)

| | Baixo esforço | Alto esforço |
|---|---|---|
| **Alto valor** | ✅ Fazer primeiro | 📅 Planejar |
| **Baixo valor** | 🤏 Se sobrar tempo | ❌ Descartar |

### 6.5 Perguntas ruins × perguntas boas

| Pergunta ruim | Problema | Pergunta reformulada |
|---|---|---|
| "Como melhorar as vendas?" | Vaga, sem ação clara | "Quais clientes têm > 70% de chance de comprar o produto X nos próximos 30 dias?" |
| "Os dados mostram algo interessante?" | Sem objetivo | "Existem segmentos de clientes com comportamento de compra distinto que justifiquem campanhas diferentes?" |
| "Vamos usar IA no RH?" | Começa pela técnica | "Quais fatores estão associados à saída voluntária de funcionários no 1º ano?" |

## 7. Analytical Approach: Identifying the Pattern to Address the Question

### 7.1 Tipos de padrão que os dados podem revelar

| Padrão | Descrição | Pergunta associada | Técnica |
|---|---|---|---|
| **Relação** | Como uma variável muda com outra | "Preço afeta demanda?" | Correlação, regressão |
| **Agrupamento** | Observações similares entre si | "Quais perfis de cliente existem?" | Clustering |
| **Classificação** | Pertencimento a categorias conhecidas | "É spam ou não?" | Classificadores |
| **Tendência / Sazonalidade** | Comportamento ao longo do tempo | "As vendas crescem no fim do ano?" | Séries temporais |
| **Anomalia** | Desvio do comportamento esperado | "Esta transação é atípica?" | Detecção de outliers |
| **Associação / Sequência** | Eventos que ocorrem juntos ou em ordem | "Quem compra A compra B?" | Apriori, análise de sequência |
| **Probabilidade / Risco** | Chance de um evento | "Qual o risco de inadimplência?" | Modelos probabilísticos |

### 7.2 Do padrão ao modelo: a lógica de decisão
1. Que **padrão** responderia à pergunta?
2. Há **variável-alvo** conhecida? → supervisionado; senão → não supervisionado.
3. A resposta é **categoria** ou **número**? → classificação ou regressão.
4. O **tempo** importa? → séries temporais / sobrevivência.
5. Que **nível de interpretabilidade** o usuário exige?
6. Que **métrica** representa sucesso para o negócio?

### 7.3 Escolha da métrica conforme o padrão

| Problema | Métricas típicas | Observação |
|---|---|---|
| Classificação | Acurácia, Precisão, Recall, F1, AUC-ROC | Em classes desbalanceadas, evite acurácia isolada |
| Regressão | MAE, RMSE, R² | MAE é mais robusto a outliers |
| Clustering | Silhueta, Davies-Bouldin | Avalie também a interpretação de negócio |
| Séries temporais | MAPE, RMSE por horizonte | Use validação temporal (sem embaralhar) |

### 7.4 Caso ICC: o padrão buscado
- Padrão: **probabilidade/risco** + **classificação**.
- Existe alvo (reinternou em 30 dias: sim/não) → **supervisionado**.
- Resposta categórica → **classificação binária**.
- Interpretabilidade alta → **árvore de decisão**.
- Métrica prioritária → **recall** (custo de falso negativo é alto).

## 8. From Requirements to Collection

### 8.1 Data Requirements (Requisitos de dados)
Com a abordagem definida, respondemos: **quais dados, em que formato, com que granularidade e de que período?**

Para o caso ICC, definiu-se:
- **Unidade de análise:** uma linha por paciente por internação (índice).
- **Conteúdo:** dados demográficos, diagnósticos (CID), procedimentos, medicamentos na alta, comorbidades, histórico de internações, dados socioeconômicos.
- **Alvo:** reinternação em ≤ 30 dias (0/1).
- **Período:** internações dos últimos 3 anos.
- **Formato:** tabular, uma coluna por variável; códigos padronizados.
- **Critérios de exclusão:** pacientes que faleceram na internação, transferências.

Checklist de requisitos:
- [ ] Conteúdo (quais variáveis)
- [ ] Formato (tabela, JSON, imagem, texto)
- [ ] Granularidade (paciente, transação, dia)
- [ ] Volume mínimo esperado
- [ ] Janela temporal
- [ ] Qualidade mínima (% de nulos aceitável)
- [ ] Restrições legais/éticas (anonimização, consentimento)

### 8.2 Data Collection (Coleta de dados)
Identificar **fontes**, obter **acesso** e **reunir** os dados.

| Tipo de fonte | Exemplos | Forma de acesso |
|---|---|---|
| Bancos transacionais | ERP, prontuário eletrônico, CRM | SQL, exportações |
| Data warehouse / lake | Snowflake, BigQuery, S3 | SQL, Spark |
| APIs | Redes sociais, clima, cotações | REST/GraphQL |
| Arquivos | CSV, Excel, PDF, logs | Leitura direta / parsing |
| Web | Sites públicos | Web scraping (respeitando termos) |
| Dados abertos | IBGE, DATASUS, data.gov | Download / API |
| Dados comprados | Bureaus de crédito, painéis | Contratos |

### 8.3 Lacunas e revisão iterativa
Ao coletar, é comum descobrir que:
- Dados desejados **não existem** → buscar proxy ou remover a variável;
- Dados existem mas são **inacessíveis** (privacidade, custo) → revisar requisitos;
- Qualidade é **insuficiente** → adiar variáveis ou planejar coleta futura;
- Dados são mais **ricos** do que o esperado → oportunidade de ampliar a pergunta.

→ Isso reforça a **iteratividade**: Data Collection realimenta Data Requirements e, às vezes, Analytic Approach.

### 9.2 Exemplo prático ponta a ponta (simplificado)
A seguir, aplicamos as etapas 1–8 (e uma prévia de Modeling/Evaluation) com **dados sintéticos** inspirados no caso ICC. O objetivo é ver a metodologia "em movimento", não obter um modelo clínico real.

````python
# ETAPAS 1-2: Business Understanding + Analytic Approach (documentação em código)
problem_statement = {
    "stakeholder": "Equipe de enfermagem e diretoria clínica",
    "dor": "Alta taxa de reinternação em 30 dias de pacientes com ICC",
    "pergunta_analitica": "O paciente será reinternado em <= 30 dias após a alta?",
    "tipo_resposta": "Binária (sim/não) + probabilidade",
    "abordagem": "Classificação supervisionada, modelo interpretável (árvore de decisão)",
    "acao": "Priorizar follow-up nos pacientes de alto risco",
    "metrica_sucesso": "Recall >= 0.75 mantendo precisão >= 0.40",
}
for k, v in problem_statement.items():
    print(f"{k:>20}: {v}")
````

````python
# ETAPAS 3-4: Data Requirements + Data Collection (simulando a coleta com dados sintéticos)
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 1500

df = pd.DataFrame({
    "idade": rng.integers(40, 95, n),
    "sexo": rng.choice(["M", "F"], n),
    "n_internacoes_12m": rng.poisson(1.2, n),
    "comorbidades": rng.integers(0, 7, n),
    "dias_internado": rng.integers(1, 21, n),
    "n_medicamentos_alta": rng.integers(2, 15, n),
})

# Geração do alvo com uma relação plausível (apenas para fins didáticos)
logit = (-4.5 + 0.03*df["idade"] + 0.6*df["n_internacoes_12m"]
         + 0.35*df["comorbidades"] + 0.05*df["dias_internado"] + 0.08*df["n_medicamentos_alta"])
prob = 1/(1+np.exp(-logit))
df["reinternou_30d"] = (rng.random(n) < prob).astype(int)

print("Formato:", df.shape)
print("Taxa de reinternação:", round(df["reinternou_30d"].mean(), 3))
df.head()
````

````python
# ETAPA 5: Data Understanding – os dados representam o problema?
print(df.describe().round(2))
print("\nValores nulos por coluna:\n", df.isna().sum())
print("\nTaxa de reinternação por nº de internações prévias:")
print(df.groupby("n_internacoes_12m")["reinternou_30d"].agg(["mean", "count"]).round(3))
````

````python
# ETAPA 6: Data Preparation – codificação e separação treino/teste
from sklearn.model_selection import train_test_split

X = pd.get_dummies(df.drop(columns="reinternou_30d"), columns=["sexo"], drop_first=True)
y = df["reinternou_30d"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
print("Treino:", X_train.shape, "| Teste:", X_test.shape)
print("Colunas:", list(X.columns))
````

````python
# ETAPAS 7-8: Modeling + Evaluation – árvore de decisão interpretável
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import classification_report, roc_auc_score

modelo = DecisionTreeClassifier(max_depth=4, min_samples_leaf=30, class_weight="balanced", random_state=42)
modelo.fit(X_train, y_train)

y_pred = modelo.predict(X_test)
y_prob = modelo.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred, digits=3))
print("AUC-ROC:", round(roc_auc_score(y_test, y_prob), 3))
print("\nRegras da árvore (interpretabilidade para os médicos):\n")
print(export_text(modelo, feature_names=list(X.columns)))
````

````python
# Ajuste de limiar conforme capacidade operacional da equipe (decisão de negócio!)
from sklearn.metrics import precision_score, recall_score

print(f"{'limiar':>7} | {'recall':>7} | {'precisao':>9} | {'% pacientes sinalizados':>24}")
for t in [0.3, 0.4, 0.5, 0.6, 0.7]:
    pred_t = (y_prob >= t).astype(int)
    print(f"{t:>7.2f} | {recall_score(y_test, pred_t):>7.3f} | {precision_score(y_test, pred_t, zero_division=0):>9.3f} | {pred_t.mean()*100:>23.1f}%")
````

### 9.3 Leitura do exemplo à luz da metodologia
- A **pergunta** (sim/não) definiu a **abordagem** (classificação) → que definiu os **requisitos de dados** (variáveis por paciente + alvo binário).
- A **interpretabilidade** exigida pelos médicos definiu a **técnica** (árvore).
- O **limiar de decisão** não é técnico: depende de quantos pacientes a enfermagem consegue acompanhar (**Business Understanding** volta a aparecer na Evaluation).
- Em produção (**Deployment**), o modelo rodaria no momento da alta; em **Feedback**, compararíamos a taxa de reinternação antes/depois e re-treinaríamos periodicamente.
